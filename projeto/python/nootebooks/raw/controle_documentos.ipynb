## Tratamento de dados

```python
df_original = df.copy()
```

```python
df.info()
```

```python
df.head()
```

```python
df['id_documento'] = df['id_documento'].astype('str')
```

```python
# removendo espacoes extras
df.columns = df.columns.str.strip()
```

```python
# campos title

campos_title = [
    'tipo_documento'
    , 'descricao'
    , 'responsavel'
    , 'status'
    , 'local_arquivo'
]
```

```python
# Padronizar letras maiusculas/minusculas
for campo in campos_title:
    df[campo] = (
      df[campo].str.title()
        
    ) 
```

```python
# campos categoricos

campos_categoricos = [
    'tipo_documento'
   , 'status'
   , 'local_arquivo'
]
```

```python
# Padronizar campos categoricos

for campo in campos_categoricos:
    df[campo] = (
      df[campo].astype('category')
        
    ) 
```

```python
df['data_recebimento'] = pd.to_datetime(df['data_recebimento'], errors='coerce', format='%Y-%m-%d')
```
