## Tratamento de dados

```python
df_original = df.copy()
```

```python
df.info()
```

```python
df.head()
```

```python
df['id_colaborador'] = df['id_colaborador'].astype('str')
df['salario'] = df['salario'].astype('float')
```

```python
# remover espaco extras
df.columns = df.columns.str.strip()
```

```python
# Padronizar letras maiusculas/minusculas
df['nome'] = df['nome'].str.title()
df['departamento'] = df['departamento'].str.title()
df['cargo'] = df['cargo'].str.title()
df['status'] = df['status'].str.title()
```

```python
# padronizar datas
df['data_admissao'] = pd.to_datetime(df['data_admissao'], errors='coerce', format='%Y-%m-%d')
```

```python
# campos categoricos
categoricos = [
    'departamento'
    , 'cargo'
    , 'status'
]
```

```python
for campo in categoricos:
    df[campo] = (
        df[campo]
        .astype('category')
    )
```
