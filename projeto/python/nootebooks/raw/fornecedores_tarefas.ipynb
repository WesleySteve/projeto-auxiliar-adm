## Tratamento de dados

```python
df_original = df.copy()
```

```python
df.info()
```

```python
df.head()
```

```python
campos_texto = [
    'id_fornecedor'
    , 'fornecedor'
    , 'cidade'
]
```

```python
for campo in campos_texto:
    df[campo] = (
        df[campo]
        .astype('str')
    )
```

```python
df.columns = df.columns.str.strip()
```

```python
campos_categoricos = [
    'categoria'
    , 'status'
]
```

```python
for campo in campos_categoricos:
    df[campo] = (
        df[campo]
        .astype('category')
    )
```
